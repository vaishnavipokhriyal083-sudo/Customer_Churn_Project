# **CUSTOMER CHURN PROJECT**


# 1. Final Model Performance Comparison.

In this project, three distinct machine learning models were developed, trained, and evaluated to predict customer churn. Based on the test metrics obtained, Logistic Regression is the absolute winning model for this dataset:

*Logistic Regression (Winner): Classification Accuracy of 82.27%*

*Random Forest Classifier: Classification Accuracy of 0.89% *

*Decision Tree Classifier: Classification Accuracy of 76.59%*

# 2. Deep-Dive Performance Insights

*Why Logistic Regression Won:* The fact that Logistic Regression achieved the highest accuracy (82.27%) indicates that the primary drivers of customer churn (such as contract type, monthly charges, and customer tenure) have a strong, clean, linear relationship with the target variable.

*Tree-Based Models:* The Decision Tree Classifier provided a lower accuracy baseline (76.59%). While upgrading to an ensemble Random Forest Classifier improved performance to 0.89%, it did not outperform the simpler linear model, likely due to slight overfitting on complex non-linear combinations that act as noise in this specific dataset.

*Model Precision:* The final model pipeline reached a Precision Score of 79.16%. This means that out of all the customers the system flags as "likely to churn," roughly 79% of them are true churn risks. High precision ensures that the business does not waste retention marketing budgets on customers who have no intention of leaving.

# 3. Final Business Recommendations

*Deployment:* The Logistic Regression pipeline should be deployed as the primary prediction engine for customer account analytics due to its superior 82.27% accuracy.

*Proactive Intervention:*  The customer success team can run live account metrics through this trained pipeline weekly. Accounts flagged by the model can automatically be targeted with personalized save-offers, contract adjustments, or loyalty outreach before they officially cancel their service.
